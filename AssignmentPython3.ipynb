```python
import random 
# Generate random number between 1 and 10 
secret_number = random.randint(1, 10)
max_attempt = 3
attempts = 0
while attempts < 3:
    num = int(input('Guess the number(between 1 to 10):'))
    if num >= 0 and num <= 10:
        if num == secret_number:
            print("**Congratzz!!You guessed the correct number**")
            break
        elif num < secret_number:
            print("**You guessed number is too Low.Please try again**")
        elif num > secret_number:
            print("**You guessed number is too High.Try again**")
    else:
        print("**Your guess is out of range.Please guess a number between 1 and 10.**")
        continue
    attempts += 1
        
else:
    print( "Better luck next time!") 
```

```python
def calculate_bmi(weight, height):
    heightper = height/100
    bmi = weight/heightper**2
    if(bmi < 18.5):
        status = 'Underweight'
    elif(18.5 <= bmi <= 24.9):
        status = 'Healthy'
    elif(25 <= bmi <= 29.9):
        status = 'Overweight'
    else:
       status = 'Obesity'
    print('!!Your BMI is:',round(bmi,2),status,'!!')
height = float(input('Enter your height(in Cm):'))
weight = float(input('Enter your weight(in Kg):'))
calculate_bmi(weight, height)
```
